# Mappa soggetti

Notebook dedicato ai soggetti della rete. Legge il JSON enriched prodotto da `s4.1.1`, usa i flag ruolo già pronti e costruisce una mappa separata per soggetti, governance ed ente capofila.

**Nota metodologica**: la governance totale è l’aggregazione di coordinamento, gestione e monitoraggio. Le note che indicano solo una partecipazione generica alla governance vengono tracciate come `governance generica`, senza forzare una delle tre componenti specifiche.


```python

from pathlib import Path
import json
import re
import pandas as pd
import folium
from folium.plugins import MarkerCluster
from IPython.display import display

INPUT_JSON = Path(r"output\data\step_3.0\3.0_risultati_enriched_merged_geo.json")
GEOJSON_REGIONI = Path(r"geo-json\limits_IT_regions.geojson")
GEOJSON_PROVINCE = Path(r"geo-json\limits_IT_provinces.geojson")
OUTPUT_MAPPA = Path(r"output\reports\3.3.3_html\3.3.3_mappa_soggetti_v3.html")
OUTPUT_DIR_TABLE = Path(r"output\reports\3.3.3_csv")
```

```python

def clean_text(x):
    if pd.isna(x):
        return ""
    return re.sub(r"\s+", " ", str(x).strip())

def clean_title(x):
    x = clean_text(x)
    return x.title() if x else ""

def to_float_or_none(x):
    try:
        return float(x)
    except Exception:
        return None

def has_valid_coords(lat, lon):
    return lat is not None and lon is not None and -90 <= lat <= 90 and -180 <= lon <= 180

def to_bool(x):
    if isinstance(x, bool):
        return x
    if isinstance(x, (int, float)):
        return x == 1
    if isinstance(x, str):
        return x.strip().lower() in {"1", "true", "yes", "si", "sì", "x"}
    return False

def iter_items(obj):
    if isinstance(obj, dict):
        if "soggetti" in obj or "entities" in obj:
            yield obj
        for v in obj.values():
            yield from iter_items(v)
    elif isinstance(obj, list):
        for item in obj:
            yield from iter_items(item)

def iter_entities(item):
    if not isinstance(item, dict):
        return []
    if isinstance(item.get("soggetti"), list):
        return [x for x in item["soggetti"] if isinstance(x, dict)]
    if isinstance(item.get("entities"), list):
        return [x for x in item["entities"] if isinstance(x, dict)]
    return []
```

```python

with open(INPUT_JSON, "r", encoding="utf-8") as f:
    merged = json.load(f)

geo_regioni = None
geo_province = None
if GEOJSON_REGIONI.exists():
    with open(GEOJSON_REGIONI, "r", encoding="utf-8") as f:
        geo_regioni = json.load(f)
if GEOJSON_PROVINCE.exists():
    with open(GEOJSON_PROVINCE, "r", encoding="utf-8") as f:
        geo_province = json.load(f)
```

```python

punti = []
for item in iter_items(merged):
    file_name = clean_text(item.get("file") or item.get("files") or "")
    for ent in iter_entities(item):
        lat = to_float_or_none(ent.get("lat"))
        lon = to_float_or_none(ent.get("lon"))
        if not has_valid_coords(lat, lon):
            continue
        ruolo_raw = ent.get("ruolo", [])
        ruolo_txt = ", ".join([clean_text(x) for x in ruolo_raw if clean_text(x)]) if isinstance(ruolo_raw, list) else clean_text(ruolo_raw)
        punti.append({
            "file": file_name,
            "nome": clean_text(ent.get("nome")),
            "tipo": clean_text(ent.get("tipo")),
            "comune": clean_title(ent.get("comune") or ent.get("comune_matchato")),
            "provincia": clean_title(ent.get("provincia")),
            "sigla_provincia": clean_text(ent.get("sigla_provincia")).upper(),
            "regione": clean_title(ent.get("regione")),
            "codice_comune": clean_text(ent.get("codice_comune")),
            "codice_provincia": clean_text(ent.get("codice_provincia")),
            "codice_regione": clean_text(ent.get("codice_regione")),
            "livello_territoriale": clean_text(ent.get("livello_territoriale")).lower(),
            "lat": round(lat, 6),
            "lon": round(lon, 6),
            "ruolo": ruolo_txt,
            "governance_detail": clean_text(ent.get("governance_detail")),
            "ente_capofila": "sì" if to_bool(ent.get("is_ente_capofila")) else "no",
            "is_firmatario": to_bool(ent.get("is_firmatario")),
            "is_proponente": to_bool(ent.get("is_proponente")),
            "is_attore": to_bool(ent.get("is_attore")),
            "is_gestione": to_bool(ent.get("is_gestione")),
            "is_monitoraggio": to_bool(ent.get("is_monitoraggio")),
            "is_coordinamento": to_bool(ent.get("is_coordinamento")),
            "is_governance": to_bool(ent.get("is_governance")),
            "is_governance_generica": to_bool(ent.get("is_governance_generica")),
            "is_ente_capofila": to_bool(ent.get("is_ente_capofila")),
        })

df_punti = pd.DataFrame(punti).copy()
if df_punti.empty:
    raise SystemExit("Nessun punto valido trovato nel JSON enriched.")
for col in ["file", "nome", "tipo", "comune", "provincia", "sigla_provincia", "regione", "codice_comune", "codice_provincia", "codice_regione", "livello_territoriale", "ruolo", "governance_detail"]:
    df_punti[col] = df_punti[col].fillna("").astype(str).str.strip()
for col in ["is_firmatario", "is_proponente", "is_attore", "is_gestione", "is_monitoraggio", "is_coordinamento", "is_governance"]:
    df_punti[col] = df_punti[col].fillna(False).astype(bool)

df_punti["nome_key"] = df_punti["nome"].str.lower()
df_punti["tipo_key"] = df_punti["tipo"].str.lower()
df_punti["comune_key"] = df_punti["comune"].str.lower()
df_punti["provincia_key"] = df_punti["provincia"].str.lower()
df_punti["regione_key"] = df_punti["regione"].str.lower()
df_punti["ruolo_key"] = df_punti["ruolo"].str.lower()
df_punti["livello_key"] = df_punti["livello_territoriale"].str.lower()

df_punti = df_punti.drop_duplicates(subset=["lat", "lon", "nome_key", "tipo_key", "comune_key", "provincia_key", "regione_key", "ruolo_key", "livello_key"]).reset_index(drop=True)

print("Punti puliti:", len(df_punti))
print("Firmatari:", int(df_punti["is_firmatario"].sum()))
print("Proponenti:", int(df_punti["is_proponente"].sum()))
print("Attori:", int(df_punti["is_attore"].sum()))
print("Gestione:", int(df_punti["is_gestione"].sum()))
print("Monitoraggio:", int(df_punti["is_monitoraggio"].sum()))
print("Coordinamento:", int(df_punti["is_coordinamento"].sum()))
print("Governance totale:", int(df_punti["is_governance"].sum()))
print("Governance generica:", int(df_punti["is_governance_generica"].sum()))
print("Enti capofila:", int(df_punti["is_ente_capofila"].sum()))
display(df_punti.head())
```

```python

df_firmatari = df_punti[df_punti["is_firmatario"]].copy()
df_proponenti = df_punti[df_punti["is_proponente"]].copy()
df_attori = df_punti[df_punti["is_attore"]].copy()
df_gestione = df_punti[df_punti["is_gestione"]].copy()
df_monitoraggio = df_punti[df_punti["is_monitoraggio"]].copy()
df_coordinamento = df_punti[df_punti["is_coordinamento"]].copy()
df_governance = df_punti[df_punti["is_governance"]].copy()
df_governance_generica = df_punti[df_punti["is_governance_generica"]].copy()
df_enti_capofila = df_punti[df_punti["is_ente_capofila"]].copy()
df_governance = df_governance.drop_duplicates(subset=["lat", "lon", "nome_key", "tipo_key", "comune_key", "provincia_key", "regione_key", "livello_key"]).copy()
```

```python

def style_regioni(feature):
    return {"fillColor": "#00000000", "color": "#cc0000", "weight": 2, "fillOpacity": 0.0}

def style_province(feature):
    return {"fillColor": "#00000000", "color": "#666666", "weight": 1, "fillOpacity": 0.0}

def make_popup(row):
    ruolo_txt = row.get("ruolo", "")
    if row.get("is_governance", False) and row.get("governance_detail", ""):
        ruolo_txt = row.get("governance_detail", "")
    return f"""
    <div style='font-size:13px;line-height:1.4;'>
      <b>Nome:</b> {row.get('nome','')}<br>
      <b>Tipo:</b> {row.get('tipo','')}<br>
      <b>Ruolo:</b> {ruolo_txt}<br>
      <b>Livello:</b> {row.get('livello_territoriale','')}<br>
      <b>Comune:</b> {row.get('comune','')}<br>
      <b>Provincia:</b> {row.get('provincia','')} {row.get('sigla_provincia','')}<br>
      <b>Regione:</b> {row.get('regione','')}<br>
      <b>File:</b> {row.get('file','')}
    </div>
    """

def add_points_layer(map_obj, df, layer_name, color="blue", cluster=True, show=False):
    feature_group = folium.FeatureGroup(name=layer_name, show=show)
    target = MarkerCluster(name=layer_name).add_to(feature_group) if cluster else feature_group
    for _, row in df.iterrows():
        folium.CircleMarker(location=[row["lat"], row["lon"]], radius=5, color=color, fill=True, fill_opacity=0.75, popup=folium.Popup(make_popup(row), max_width=450), tooltip=row.get("nome", "")).add_to(target)
    feature_group.add_to(map_obj)
```

```python

centro_lat = df_punti["lat"].mean()
centro_lon = df_punti["lon"].mean()
m = folium.Map(location=[centro_lat, centro_lon], zoom_start=6, tiles="CartoDB positron")
if geo_regioni is not None:
    folium.GeoJson(geo_regioni, name="Confini regioni", style_function=style_regioni).add_to(m)
if geo_province is not None:
    folium.GeoJson(geo_province, name="Confini province", style_function=style_province, show=False).add_to(m)

add_points_layer(m, df_punti, "Tutti i soggetti", color="gray", cluster=True, show=True)
add_points_layer(m, df_firmatari, "Firmatari", color="blue", cluster=True, show=False)
add_points_layer(m, df_proponenti, "Proponenti", color="green", cluster=True, show=False)
add_points_layer(m, df_attori, "Attori", color="orange", cluster=True, show=False)
add_points_layer(m, df_governance, "Governance totale", color="red", cluster=True, show=True)
add_points_layer(m, df_gestione, "Gestione", color="purple", cluster=True, show=False)
add_points_layer(m, df_monitoraggio, "Monitoraggio", color="cadetblue", cluster=True, show=False)
add_points_layer(m, df_coordinamento, "Coordinamento", color="darkred", cluster=True, show=False)
folium.LayerControl(collapsed=False).add_to(m)
m
```

```python

OUTPUT_MAPPA.parent.mkdir(parents=True, exist_ok=True)
OUTPUT_DIR_TABLE.mkdir(parents=True, exist_ok=True)
m.save(OUTPUT_MAPPA)
df_punti.to_csv(OUTPUT_DIR_TABLE / "punti_soggetti_v3.csv", index=False, encoding="utf-8-sig")
df_governance.to_csv(OUTPUT_DIR_TABLE / "punti_governance_v3.csv", index=False, encoding="utf-8-sig")
print(f"Mappa salvata in: {OUTPUT_MAPPA}")
print(f"CSV salvati in: {OUTPUT_DIR_TABLE}")
```
